## 8. Data Cleaning

Based on the data understanding phase, we identified:
- **Rainfall_mm**: 48 missing values (1.2%)
- **Soil_Moisture_pct**: 40 missing values (1.0%)
- **Yield_Tonnes_Ha**: 32 missing values (0.8%)
- **No duplicate rows**
- **No invalid data types**

**Cleaning decisions:**
1. Fill missing numerical values with the **median** of their respective columns (robust to outliers)
2. No rows need to be dropped since missing values are few and can be imputed
3. The original CSV is not modified — all cleaning is done on the working DataFrame

```python
# Store original counts for verification
original_shape = df.shape
missing_before = df.isnull().sum().sum()

# Fill missing values with median
columns_with_missing = ['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']
for col in columns_with_missing:
    median_val = df[col].median()
    count_missing = df[col].isnull().sum()
    df[col] = df[col].fillna(median_val)
    print(f"Filled {count_missing} missing values in '{col}' with median = {median_val}")

missing_after = df.isnull().sum().sum()
print(f"\nMissing values before cleaning: {missing_before}")
print(f"Missing values after cleaning: {missing_after}")
print(f"Dataset shape unchanged: {df.shape == original_shape}")
```

### 8.1 Verify Data After Cleaning

```python
# Confirm no missing values remain
print("Missing values after cleaning:")
print(df.isnull().sum()[df.isnull().sum() > 0])
if df.isnull().sum().sum() == 0:
    print("\n✓ No missing values remain. Dataset is clean.")
```

## 10. Season-wise Analysis

### 10.1 Season Distribution

```python
season_counts = df['Season'].value_counts()
print("Number of records per season:")
print(season_counts)
print(f"\nTotal: {season_counts.sum()}")
```

### 10.2 Average Yield by Season

```python
season_yield = df.groupby('Season')['Yield_Tonnes_Ha'].agg(['mean', 'median', 'std']).round(2)
season_yield.columns = ['Mean Yield (t/ha)', 'Median Yield (t/ha)', 'Std Dev']
print(season_yield)

# Chart 1: Average Yield by Season
fig, ax = plt.subplots(figsize=(8, 5))
colors = ['#2E86AB', '#A23B72', '#F18F01']
bars = ax.bar(season_yield.index, season_yield['Mean Yield (t/ha)'], color=colors, edgecolor='black', linewidth=0.5)
ax.set_title('Average Yield by Season', fontweight='bold', pad=15)
ax.set_xlabel('Season')
ax.set_ylabel('Average Yield (Tonnes/Hectare)')
for bar, val in zip(bars, season_yield['Mean Yield (t/ha)']):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1, f'{val}', ha='center', fontsize=11)
plt.tight_layout()
plt.savefig('charts/01_avg_yield_by_season.png', dpi=150, bbox_inches='tight')
plt.show()
print("\nChart saved: charts/01_avg_yield_by_season.png")
```

**Observation:** Kharif season shows the highest average yield, followed by Rabi and Zaid.

**Interpretation:** This suggests that Kharif season conditions (higher rainfall, monsoon period) are associated with higher average yields in this dataset. However, the differences should be considered alongside the variation (standard deviation) within each season.

### 10.3 Average Profit by Season

```python
season_profit = df.groupby('Season')['Profit_INR'].agg(['mean', 'median']).round(2)
season_profit.columns = ['Mean Profit (₹)', 'Median Profit (₹)']
print(season_profit)

# Profitable vs loss-making farms
print("\nProfitable vs Loss-making farms:")
for season in ['Kharif', 'Rabi', 'Zaid']:
    sdf = df[df['Season'] == season]
    profitable = (sdf['Profit_INR'] > 0).sum()
    loss = (sdf['Profit_INR'] <= 0).sum()
    pct = round(profitable / len(sdf) * 100, 1)
    print(f"  {season}: {profitable} profitable ({pct}%), {loss} loss-making")

# Chart 2: Average Profit by Season
fig, ax = plt.subplots(figsize=(8, 5))
profit_vals = season_profit['Mean Profit (₹)']
bar_colors = ['#2E86AB' if v >= 0 else '#E63946' for v in profit_vals]
bars = ax.bar(profit_vals.index, profit_vals.values, color=bar_colors, edgecolor='black', linewidth=0.5)
ax.set_title('Average Profit by Season', fontweight='bold', pad=15)
ax.set_xlabel('Season')
ax.set_ylabel('Average Profit (₹)')
ax.axhline(y=0, color='black', linewidth=0.8, linestyle='--')
for bar, val in zip(bars, profit_vals.values):
    ypos = bar.get_height() + 3000 if val >= 0 else bar.get_height() - 12000
    ax.text(bar.get_x() + bar.get_width()/2, ypos, f'₹{val:,.0f}', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('charts/02_avg_profit_by_season.png', dpi=150, bbox_inches='tight')
plt.show()
print("\nChart saved: charts/02_avg_profit_by_season.png")
```

**Observation:** Kharif has the highest average profit (₹1,78,915), Rabi is moderate (₹87,689), and Zaid shows a negative average profit (-₹24,805).

**Interpretation:** Kharif season appears to be the most profitable on average. Zaid season's negative average profit indicates that more farms in this season incur losses than make profits in this dataset. This may be associated with the lower rainfall and different crop mix during Zaid.

### 10.4 Total Production by Season

```python
season_production = df.groupby('Season')['Production_Tonnes'].agg(['sum', 'mean']).round(2)
season_production.columns = ['Total Production (Tonnes)', 'Mean Production (Tonnes)']
print(season_production)

# Chart 3: Total Production by Season
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(season_production.index, season_production['Total Production (Tonnes)'], 
              color=['#2E86AB', '#A23B72', '#F18F01'], edgecolor='black', linewidth=0.5)
ax.set_title('Total Production by Season', fontweight='bold', pad=15)
ax.set_xlabel('Season')
ax.set_ylabel('Total Production (Tonnes)')
for bar, val in zip(bars, season_production['Total Production (Tonnes)']):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 500, f'{val:,.0f}', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('charts/03_total_production_by_season.png', dpi=150, bbox_inches='tight')
plt.show()
print("\nChart saved: charts/03_total_production_by_season.png")
```

## 15. Correlation Analysis

We examine correlations between key numerical variables to identify relationships.

**Important note:** Correlation measures the strength of a linear association between two variables. It does not imply causation.

```python
# Select meaningful numerical columns for correlation
corr_cols = ['Rainfall_mm', 'Avg_Temperature_C', 'Humidity_pct', 'Soil_pH', 
             'Soil_Moisture_pct', 'Fertilizer_kg_ha', 'Yield_Tonnes_Ha', 
             'Production_Tonnes', 'Profit_INR', 'Water_Used_m3', 
             'Water_Efficiency_t_per_1000m3', 'Disease_Pest_Risk_pct']

corr_matrix = df[corr_cols].corr().round(3)

# Chart 12: Correlation Heatmap
fig, ax = plt.subplots(figsize=(12, 10))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r', 
            center=0, vmin=-1, vmax=1, square=True, linewidths=0.5,
            cbar_kws={'shrink': 0.8}, ax=ax, annot_kws={'size': 8})
ax.set_title('Correlation Heatmap — Key Agricultural Variables', fontweight='bold', pad=20)
plt.tight_layout()
plt.savefig('charts/12_correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print("Chart saved: charts/12_correlation_heatmap.png")
```

### 15.1 Notable Correlations

```python
# Extract notable correlations (|r| > 0.3)
print("Notable correlations (|r| > 0.3):\n")
seen = set()
for i in range(len(corr_matrix)):
    for j in range(i+1, len(corr_matrix)):
        r = corr_matrix.iloc[i, j]
        if abs(r) > 0.3:
            pair = (corr_matrix.index[i], corr_matrix.columns[j])
            if pair not in seen:
                seen.add(pair)
                direction = "positive" if r > 0 else "negative"
                strength = "strong" if abs(r) > 0.7 else "moderate"
                print(f"  {pair[0]} ↔ {pair[1]}: r = {r:.3f} ({strength} {direction})")
```

**Key correlation findings:**
- **Yield and Production** show a strong positive association (r ≈ 0.885), which is expected since production depends on yield
- **Yield and Water Efficiency** are strongly correlated (r ≈ 0.915), as water efficiency is calculated from yield
- **Rainfall and Disease Risk** show a moderate positive association (r ≈ 0.625), suggesting higher rainfall is linked to greater disease risk
- **Rainfall and Humidity** are moderately correlated (r ≈ 0.524), as expected
- **Rainfall and Soil Moisture** show a moderate positive association (r ≈ 0.516)

**Important:** These correlations show associations, not causation. For example, rainfall is associated with disease risk, but other factors may contribute.
